```python
engagement_cols = [
    "jobinvolvement",
    "jobsatisfaction",
    "environmentsatisfaction",
    "relationshipsatisfaction"
]

df["engagement_index"] = df[engagement_cols].mean(axis=1)
```

```python
df["engagement_index"].describe()
```

```python
df["engagement_index"].head()
```

```python
df["burnout_score"] = 0

df.loc[df["overtime"] == "Yes", "burnout_score"] += 1

df.loc[df["worklifebalance"] <= 2, "burnout_score"] += 1
```

```python
def burnout_level(score):
    if score == 0:
        return "Low"
    elif score == 1:
        return "Medium"
    else:
        return "High"

df["burnout_risk"] = df["burnout_score"].apply(burnout_level)
```

```python
df
```

```python
overtime_analysis = df.groupby("overtime")["engagement_index"].mean()

print(overtime_analysis)
```

```python
df.groupby("overtime")[["engagement_index", "attrition"]].mean()
```

```python
travel_analysis = df.groupby("businesstravel")["engagement_index"].mean()

print(travel_analysis)
```

```python
df.groupby("businesstravel")[["engagement_index", "burnout_score"]].mean()
```

```python
median_distance = df["distancefromhome"].median()

df["commute_type"] = np.where(
    df["distancefromhome"] > median_distance,
    "Long",
    "Short"
)
```

```python
df.groupby("commute_type")["engagement_index"].mean()
```

```python
df.groupby("joblevel")["engagement_index"].mean()
```

```python
df.groupby("yearsatcompany")["engagement_index"].mean()
```

```python
bins = [0, 2, 5, 10, 40]
labels = ["0-2", "3-5", "6-10", "10+"]

df["tenure_group"] = pd.cut(
    df["yearsatcompany"],
    bins=bins,
    labels=labels
)
df.groupby("tenure_group")["engagement_index"].mean()
```

```python
bins = [0, 2, 5, 10, 20]
labels = ["0-2", "3-5", "6-10", "10+"]

df["role_tenure_group"] = pd.cut(
    df["yearsincurrentrole"],
    bins=bins,
    labels=labels
)
df.groupby("role_tenure_group")["engagement_index"].mean()
```

```python
df.groupby("attrition")[["engagement_index", "burnout_score"]].mean()
```
